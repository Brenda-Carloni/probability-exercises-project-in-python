## Ejercicio 1 

Dos dados se lanzan una vez y se observa el total obtenido. Usa una simulación para encontrar la probabilidad estimada de que la puntuación sea mayor a 7 o un número par. Una simulación es una repetición del mismo experimento multiples veces para observar su comportamiento:

- Ejecuta el experimento 1000 veces (lanza 2 dados 1000 veces, y suma el número de ambos dados).
- Lleva ua cuenta de los números y las veces que la suma fue mayor a 7 o un número par.
- Divide el número del paso 2 entre el número de iteraciones(1000).

```python
# TODO
import random
import numpy as np

#Vamos a hacer una función que devuelva un número aleatorio entre el 1 y el 6.

def tirar_un_dado():
    return random.randint(1,6)

print(tirar_un_dado()) 
```

```python
def suma_tirar_dos_dados():
    return tirar_un_dado() + tirar_un_dado()

print(suma_tirar_dos_dados())
```

```python
def simulacion(n_simulaciones = 1000):
    cuenta = 0

    for i in range(n_simulaciones):
        resultado = suma_tirar_dos_dados()
        if (resultado%2 == 0) or (resultado>7):
            cuenta += 1

    return cuenta/n_simulaciones

print(f'La probabilidad de tirar dos dados y obtener un número mayor que 7 o par es : {np.round(simulacion()*100,2)%}')
```

## Ejercicio 2

Una caja contiene 10 bolas blancas, 20 bolas rojas y 30 bolas verdes. Si tomamos 5 bolas de la caja reemplazándolas (tomando una bola, anotando el color y luego regresandola a la caja). Queremos saber la probabilidad de:

1. Tomar 3 blancas y 2 rojas.
2. Toas todas del mismo color.

Ejecuta el experimento 1000 veces y calcula las probabilidades que mencionadas.

```python
ball_box = {} #Las bolas van a ser un diccionario

# Crea la cajas con las bolas
for i in range(60):
    if i < 10:
        ball_box[i] = "White"
    elif (i > 9) and (i < 30):
        ball_box[i] = "Red"
    else:
        ball_box[i] = "Green"

print(ball_box)
            
# TODO
```

```python
def numero_de_bola():
    return random.randint(0,59) #59 porque son los números que tiene mi diccionario. Eso me va a devolver el número de la bola.

def sacar_5_bolas():
    bolas = [] # Mi lista vacía
    for i in range(5):
        numero = numero_de_bola()
        bolas.append(ball_box[numero])
    return bolas

print(sacar_5_bolas())

# Y así, cada vez que lo ejecutemos vamos a ir sacando 5 bolas. Hasta acá tenemos la caja de bolas, el número para las bolas y las 5 bolas. Ahora tenemos que hacer la simulación.
```

```python
def simulacion2(n_iteraciones = 1000):
    # Nos piden 2 cuentas:
    cuenta1= 0 #que tenga 3 blancas y 2 rojas
    cuenta2= 0 #que sean todas del mismo color

    for i in range(n_iteraciones):
        suerte = sacar_5_bolas()
        #luego de que saco 5 bolas, lo convertimos en un array:
        suerte= np.array(suerte)
        
        #Ahora necesito hacer las sumas y contar:
        bolas_blancas = sum(suerte == 'White')
        bolas_rojas = sum(suerte == 'Red')
        bolas_verdes = sum(suerte == 'Green')

        if (bolas_blancas == 3) and (bolas_rojas ==2):
            cuenta1 += 1
        if(bolas_blancas ==5) or (bolas_rojas == 5) or (bolas_verdes == 5):
            cuenta2 += 1

    return cuenta1/n_iteraciones, cuenta2/n_iteraciones

probabilidades= simulacion2()
print(f'La probabilidad de 3 blancas y 2 rojas es {np.round(probabilidades[0]* 100,2)%}')
print(f'La probabilidad de todas del mismo color es {np.round(probabilidades[1]* 100,2)%}')
```
